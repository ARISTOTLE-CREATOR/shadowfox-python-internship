## 4. If Conditions

```python
# Program 1 — BMI Category

height = float(input("Enter height in meters: "))
weight = float(input("Enter weight in kilograms: "))

if height <= 0 or weight < 0:
    print("Please enter valid positive height and non-negative weight.")
else:
    bmi = weight / (height ** 2)

    print("BMI:", round(bmi, 2))

    if bmi >= 30:
        print("Category: Obesity")
    elif bmi >= 25:
        print("Category: Overweight")
    elif bmi >= 18.5:
        print("Category: Normal")
    else:
        print("Category: Underweight")
```

```python
# Programs 2 and 3 — City to Country / Same Country

Australia = ["Sydney", "Melbourne", "Brisbane", "Perth"]
UAE = ["Dubai", "Abu Dhabi", "Sharjah", "Ajman"]
India = ["Mumbai", "Bangalore", "Chennai", "Delhi"]

def find_country(city):
    city_normalized = city.strip().casefold()

    country_lists = {
        "Australia": Australia,
        "UAE": UAE,
        "India": India
    }

    for country, cities in country_lists.items():
        if any(city_normalized == item.casefold() for item in cities):
            return country

    return None

# Program 2
city = input("Enter a city name: ").strip()
country = find_country(city)

if country:
    print(f"{city} is in {country}")
else:
    print("City not found in the given lists.")

# Program 3
city1 = input("Enter the first city: ").strip()
city2 = input("Enter the second city: ").strip()

country1 = find_country(city1)
country2 = find_country(city2)

if country1 is None or country2 is None:
    print("One or both cities were not found.")
elif country1 == country2:
    print(f"Both cities are in {country1}")
else:
    print("They don't belong to the same country")
```

## 5. For Loop

```python
# Program 1 — Roll a six-sided die at least 20 times

import random

number_of_rolls = 20
count_six = 0
count_one = 0
count_two_sixes_in_a_row = 0

previous_roll = None
rolls = []

for _ in range(number_of_rolls):
    roll = random.randint(1, 6)
    rolls.append(roll)

    if roll == 6:
        count_six += 1

    if roll == 1:
        count_one += 1

    if previous_roll == 6 and roll == 6:
        count_two_sixes_in_a_row += 1

    previous_roll = roll

print("Rolls:", rolls)
print("Number of times rolled 6:", count_six)
print("Number of times rolled 1:", count_one)
print("Number of times rolled two 6s in a row:", count_two_sixes_in_a_row)
```

```python
# Program 2 — 100 jumping jacks

total_jumping_jacks = 100
completed = 0

for set_number in range(1, 11):
    completed += 10
    remaining = total_jumping_jacks - completed

    print(f"\nSet {set_number}: You completed 10 jumping jacks.")

    if remaining == 0:
        print("Congratulations! You completed the workout.")
        break

    print(f"{remaining} jumping jacks are remaining.")

    tired = input("Are you tired? (yes/no): ").strip().lower()

    if tired in ("yes", "y"):
        skip = input(
            "Do you want to skip the remaining sets? (yes/no): "
        ).strip().lower()

        if skip in ("yes", "y"):
            print(f"You completed a total of {completed} jumping jacks.")
            break

    elif tired in ("no", "n"):
        print(f"Continue. {remaining} jumping jacks remaining.")

    else:
        print("Invalid response. Continuing the workout.")
```

```python
# Web Scraper — extraction + storage + error handling

import requests
from bs4 import BeautifulSoup
import pandas as pd

# Public practice page.
# Replace with the exact ShadowFox practice URL if one was supplied to you.
TARGET_URL = "https://books.toscrape.com/"

def scrape_webpage(url):
    try:
        response = requests.get(
            url,
            timeout=15,
            headers={"User-Agent": "Mozilla/5.0"}
        )
        response.raise_for_status()

        soup = BeautifulSoup(response.text, "html.parser")

        rows = []

        # Extract book title and price from the practice site.
        for article in soup.select("article.product_pod"):
            title_tag = article.select_one("h3 a")
            price_tag = article.select_one(".price_color")

            rows.append({
                "title": title_tag.get("title", "").strip() if title_tag else "",
                "price": price_tag.get_text(strip=True) if price_tag else ""
            })

        data = pd.DataFrame(rows)
        return data

    except requests.RequestException as error:
        print("Network/request error:", error)
        return pd.DataFrame(columns=["title", "price"])

    except Exception as error:
        print("Unexpected scraping error:", error)
        return pd.DataFrame(columns=["title", "price"])


scraped_data = scrape_webpage(TARGET_URL)

print("Records extracted:", len(scraped_data))
display(scraped_data.head(10))

# Data storage
scraped_data.to_csv("scraped_data.csv", index=False)

print("Saved scraped data to: scraped_data.csv")
```

## Hangman — word guessing game with visual progress and hints

```python
import random

WORDS_WITH_HINTS = {
    "python": "A popular programming language",
    "database": "Stores organized data",
    "algorithm": "A step-by-step procedure for solving a problem",
    "computer": "An electronic machine used for computation",
    "programming": "Writing instructions for computers",
    "internship": "A temporary learning/work experience"
}

HANGMAN_STAGES = [
r"""
  +---+
      |
      |
      |
     ===
""",
r"""
  +---+
  O   |
      |
      |
     ===
""",
r"""
  +---+
  O   |
  |   |
      |
     ===
""",
r"""
  +---+
  O   |
 /|   |
      |
     ===
""",
r"""
  +---+
  O   |
 /|\  |
      |
     ===
""",
r"""
  +---+
  O   |
 /|\  |
 /    |
     ===
""",
r"""
  +---+
  O   |
 /|\  |
 / \  |
     ===
"""
]

def play_hangman():
    word = random.choice(list(WORDS_WITH_HINTS.keys()))
    hint = WORDS_WITH_HINTS[word]

    guessed_letters = set()
    wrong_guesses = 0
    max_attempts = len(HANGMAN_STAGES) - 1

    print("=== HANGMAN ===")
    print("Hint:", hint)

    while wrong_guesses < max_attempts:
        display_word = " ".join(
            letter if letter in guessed_letters else "_"
            for letter in word
        )

        print(HANGMAN_STAGES[wrong_guesses])
        print("Word:", display_word)
        print(f"Wrong guesses: {wrong_guesses}/{max_attempts}")
        print(
            "Guessed letters:",
            ", ".join(sorted(guessed_letters)) or "None"
        )

        if all(letter in guessed_letters for letter in word):
            print("Congratulations! You guessed:", word)
            return True

        guess = input("Guess one letter: ").strip().lower()

        if len(guess) != 1 or not guess.isalpha():
            print("Please enter exactly one alphabetic character.")
            continue

        if guess in guessed_letters:
            print("You already guessed that letter.")
            continue

        guessed_letters.add(guess)

        if guess in word:
            print("Correct guess!")
        else:
            wrong_guesses += 1
            print("Wrong guess!")

    print(HANGMAN_STAGES[wrong_guesses])
    print("Game over! The word was:", word)
    return False


def run_hangman():
    while True:
        play_hangman()

        again = input("Play again? (yes/no): ").strip().lower()
        if again not in ("yes", "y"):
            print("Thanks for playing!")
            break

# Uncomment to play:
# run_hangman()
```

# Task 3 — Advanced: Jupyter Notebook Data Analysis

## Chosen project

**Jupyter Notebook Data Analysis Project**

### Research question

**How do flower measurements differ between Iris species, and which measurements provide the clearest separation?**

The project includes dataset selection, exploration, data-quality checks, research question formulation, visualization, analysis, and conclusions.


```python
# Dataset selection and loading

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=[
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ]
)

df["species"] = [iris.target_names[target] for target in iris.target]

print("Dataset shape:", df.shape)
display(df.head())
```

```python
# Dataset exploration

print("Columns:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes)

print("\nSpecies distribution:")
display(df["species"].value_counts())

print("\nDescriptive statistics:")
display(df.describe())
```

```python
# Data cleaning / quality checks

print("Missing values:")
display(df.isnull().sum())

print("Duplicate rows:", df.duplicated().sum())

# Remove exact duplicate rows if any exist.
df_clean = df.drop_duplicates().copy()

print("Shape after duplicate check:", df_clean.shape)
```

```python
# Research question support — group means

group_means = df_clean.groupby("species")[
    ["sepal_length", "sepal_width", "petal_length", "petal_width"]
].mean()

display(group_means)
```

```python
# Visualization 1 — Petal length by species

df_clean.boxplot(column="petal_length", by="species")
plt.title("Petal Length by Iris Species")
plt.suptitle("")
plt.xlabel("Species")
plt.ylabel("Petal Length")
plt.show()
```

```python
# Visualization 2 — Petal width by species

df_clean.boxplot(column="petal_width", by="species")
plt.title("Petal Width by Iris Species")
plt.suptitle("")
plt.xlabel("Species")
plt.ylabel("Petal Width")
plt.show()
```

```python
# Visualization 3 — Petal length vs petal width

for species in df_clean["species"].unique():
    subset = df_clean[df_clean["species"] == species]

    plt.scatter(
        subset["petal_length"],
        subset["petal_width"],
        label=species
    )

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.legend()
plt.show()
```

```python
# Visualization 4 — Feature correlations

numeric_columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

correlation = df_clean[numeric_columns].corr()
display(correlation)

plt.figure(figsize=(7, 5))
plt.imshow(correlation, aspect="auto")
plt.xticks(range(len(numeric_columns)), numeric_columns, rotation=45)
plt.yticks(range(len(numeric_columns)), numeric_columns)
plt.colorbar(label="Correlation")
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
# Optional additional analysis — compare the range of each feature by species

summary = df_clean.groupby("species")[numeric_columns].agg(["mean", "min", "max"])
display(summary)
```
